```python
from datasets import load_dataset

# Login using e.g. `huggingface-cli login` to access this dataset
ds = load_dataset("jhan21/amazon-beauty-reviews-dataset")
```

```python
ds
```

```python
# download vader_lexicon using nltk.download()
from nltk.sentiment.vader import SentimentIntensityAnalyzer
import pandas as pd
import numpy as np
```

```python
#fileName = "esteebrandsdata.csv"
column = "text"
Data = ds['train'].to_pandas()
Data = Data.replace(np.nan,' ',regex=True)
sentences = list(Data[column])
```

```python
sid = SentimentIntensityAnalyzer()
sentiments = []
for sentence in sentences:
    ss = sid.polarity_scores(sentence)
    sentiments.append(ss)
sentiments_df = pd.DataFrame(sentiments)
```

```python
sentiments_df
```

```python
sentiments_df.insert(0, "text", Data[column].values)
```

```python
def get_sentiment(compound):
    if compound >= 0.05:
        return "Positive"
    elif compound <= -0.05:
        return "Negative"
    else:
        return "Neutral"

sentiments_df["sentiment"] = sentiments_df["compound"].apply(get_sentiment)

sentiments_df.head()
```

```python
sentiments_df.to_csv("sentiments.csv", index=False)
```
