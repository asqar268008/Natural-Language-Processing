```python
import pandas as pd
text = pd.read_table('Dataset/A.P.J.Abdul Kalam.txt',header=None,encoding='utf-8')
text = text[0][0]
print(text)
```

```python
from nltk.tokenize import word_tokenize
token = word_tokenize(text)
token
```

```python
from nltk import pos_tag
pos = pos_tag(token)
pos
```

```python
import nltk
chunkGram = r"""Date: {<CD>?<NNP><CD>}
Name: {<NNP>+}"""
chunkParser = nltk.RegexpParser(chunkGram)
chunked = chunkParser.parse(pos)
print(chunked)
```

```python
from nltk.sem import relextract
pairs = relextract.tree2semi_rel(chunked)
print(pairs)
```
