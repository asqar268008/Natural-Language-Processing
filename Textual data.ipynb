```python
import nltk
from nltk import *
```

### Tokens
#### A meaningful unit of text is a token. Words are usually considered as tokens in NLP. The process of breaking a text based on the token is called tokenization.

```python
text = "India is a republic nation. We are proud Indians"
print(len(text))
print(text[0:5])
```

```python
token = nltk.word_tokenize(text)
token
```

### Vocabulary
#### The vocabulary of a text is the set of all unique tokens used in it.

```python
vocab = sorted(set(token))
vocab
```

### Punctuation
#### Punctuation refers to symbols used  to separate sentences and their elements and to clarify meaning.

```python
from string import punctuation
vocab_no_pun = []
for s in vocab:
    if s not in punctuation:
        vocab_no_pun.append(s)
vocab_no_pun
```

### Part of speech(POS)
#### Part of speech(POS) refers to the category to which a word is assigned based on its function. You may recall that the English language has 8 parts of speech - noun, verb, adjective, adverb, pronoun, determiner, preposition, conjunction, and interjection

```python
pos_list = pos_tag(vocab_no_pun)
pos_list
```

### Root of a word - Stemming
#### Stemming is a technique used to find the root form of a word. In the root form a word is devoid of any affixes (suffixes and prefixes)

```python
from nltk.stem.snowball import SnowballStemmer
stemObj = SnowballStemmer("english")
stemObj.stem("Studying")
```

```python
stemObj.stem("Running")
```

```python
stemmed_vocab = []
for i in vocab_no_pun:
    stemmed_vocab.append(stemObj.stem(i))
stemmed_vocab
```

### Base of a word - Lemmatization
#### Lemmatization removes inflection and reduces the word to its base form

```python
from nltk.stem.wordnet import WordNetLemmatizer
lemmaObj = WordNetLemmatizer()
lemmaObj.lemmatize("went",pos='v')
```

```python
lemmaObj.lemmatize("came",pos='v')
```

```python
lemmat_vocab = []
for i in vocab_no_pun:
    lemmat_vocab.append(lemmaObj.lemmatize(i, pos='v'))
lemmat_vocab
```

### Stop words
#### Stop words are typically the most commonly occurring words in text like 'the', 'and', 'is', etc.

#### NLTK provides a pre-defined set of stopwords for English, as shown


```python
from nltk.corpus import stopwords
wo_stop_words = []
stop_words_set = set(stopwords.words("english"))
for i in vocab_no_pun:
    if i not in stop_words_set:
        wo_stop_words.append(i)
wo_stop_words
```
